```python
import pandas as pd
import re

def clean_currency_column(series: pd.Series) -> pd.Series:
    return (
        series.astype(str)
        .str.replace(r'[\"\']', '', regex=True)                 # remove quotes
        .str.replace(r'\s*USD\s*', '', regex=True, case=False)  # remove USD
        .str.replace(',', '', regex=True)                       # remove commas
        .str.replace(r'\(', '-', regex=True)                    # handle (negatives)
        .str.replace(r'\)', '', regex=True)                     # remove )
        .str.replace(r'[^0-9\.-]', '', regex=True)              # keep only numbers, dot, minus
        .replace('', '0')
        .astype(float)
        .abs()                                                  # convert to absolute
    )

# Apply to your columns
# df['bill_item_quantity'] = clean_currency_column(df['bill_item_quantity'])
# df['bill_item_price_bcy'] = clean_currency_column(df['bill_item_price_bcy'])
# df['bill_item_total_bcy'] = clean_currency_column(df['bill_item_total_bcy'])
df['ocean_freight'] = clean_currency_column(df['ocean_freight'])
df['purchase_price'] = clean_currency_column(df['purchase_price'])
df['sales_price'] = clean_currency_column(df['sales_price'])
```

```python
def fill_unknown_for_object_cols(df: pd.DataFrame) -> pd.DataFrame:
    for col in df.select_dtypes(include=['object']).columns:
        df[col] = df[col].replace(r'^\s*$', 'unknown', regex=True)
    return df

df = fill_unknown_for_object_cols(df)
```
